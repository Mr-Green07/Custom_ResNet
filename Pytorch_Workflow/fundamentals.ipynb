```python
##reshape, sqeeze, unsqueeze, 
a = tr.arange(1., 10., 1.)
a.shape
a_reshape = a.reshape(3, 3)
print(a_reshape)
a_reshape.shape
```

```python
##view
s = a.view(3, 3)
s
r = s[:, 0] = 6
print(s, a)
```

```python
##stack tensor on top

s_stack = tr.stack([s, s, s], dim=2)
print(s_stack)
```

```python
##squeeze and unsqueeze
x = tr.tensor([1,2,3,4,5,6,7])
print(x)
print(x.shape)
f = tr.squeeze(x)
print(f)
print(f.shape)
h = tr.unsqueeze(f, dim=0)
print(h)
print(h.shape)
```

```python
##permute

x = tr.rand(224, 224, 3)
print(x)
x_permuted = tr.permute(x, (2, 0, 1))
print(x_permuted)
print(x_permuted.shape)

```
